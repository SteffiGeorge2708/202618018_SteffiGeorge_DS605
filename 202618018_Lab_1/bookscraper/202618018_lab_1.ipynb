```python
import pandas as pd

df = pd.read_csv("books.csv")
import re

# Remove extra spaces and newline characters
text_columns = ["Title", "Category", "Availability", "Description"]

for col in text_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("\n", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

# Convert Price to numeric
df["Price"] = df["Price"].str.replace("£", "", regex=False).astype(float)

# Convert Rating to numbers
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating"] = df["Rating"].map(rating_map)

# Extract stock count
df["Stock_Count"] = (
    df["Availability"]
    .str.extract(r"(\d+)")
    .fillna(0)
    .astype(int)
)

df.head()
```

## Visualization 4: Average Price by Category

```python
avg_price = df.groupby("Category")["Price"].mean().sort_values()

plt.figure(figsize=(10,6))
plt.barh(avg_price.index, avg_price.values)
plt.title("Average Price by Category")
plt.xlabel("Average Price (£)")
plt.ylabel("Category")
plt.show()
```

```python
stock = df.groupby("Category")["Stock_Count"].sum().sort_values()

plt.figure(figsize=(10,6))
plt.barh(stock.index, stock.values)
plt.title("Total Stock Available by Category")
plt.xlabel("Stock Count")
plt.ylabel("Category")
plt.show()
```

# Visualization 6: Word Cloud of Book Descriptions

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Combine all book descriptions into one string
text = " ".join(df["Description"].dropna())

# Generate the word cloud
wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white"
).generate(text)

# Display it
plt.figure(figsize=(12,6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Book Descriptions")
plt.show()
```

## Feature Engineering 1: Description Word Count

```python
# Count the number of words in each book description
df["Description_Word_Count"] = (
    df["Description"]
    .fillna("")
    .astype(str)
    .apply(lambda x: len(x.split()))
)

df[["Description", "Description_Word_Count"]].head()
```

## Feature Engineering 2: Price Band

```python
# Create price categories
df["Price_Band"] = pd.cut(
    df["Price"],
    bins=[0, 20, 40, 60],
    labels=["Low", "Medium", "High"]
)

df[["Price", "Price_Band"]].head()
```

## Feature Engineering 3: Recommended Books

```python
# Recommend books with rating >=4 and price below average
average_price = df["Price"].mean()

df["Recommended"] = (
    (df["Rating"] >= 4) &
    (df["Price"] < average_price)
)

df[["Title", "Rating", "Price", "Recommended"]].head()
```

## Save Cleaned Dataset
