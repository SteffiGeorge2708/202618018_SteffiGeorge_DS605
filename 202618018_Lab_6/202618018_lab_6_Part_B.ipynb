```python
import pandas as pd
import numpy as np
import time

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load the dataset
df = pd.read_csv("emails.csv")

# Display the first 5 rows
print(df.head())

# Display the shape
print("\nDataset shape:", df.shape)

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Target distribution:")
print(df["Prediction"].value_counts())

print("\nTarget percentage:")
print(df["Prediction"].value_counts(normalize=True) * 100)
```

```python
# Columns containing word frequencies
word_columns = [
    col for col in df.columns
    if col not in ["Email No.", "Prediction", "clean_text"]
]

print("Number of word-frequency columns:", len(word_columns))
print("\nFirst 20 word columns:")
print(word_columns[:20])
```

```python
def create_text(row):
    words = []

    for word in word_columns:
        count = int(row[word])

        if count > 0:
            words.extend([word] * count)

    return " ".join(words)
```

```python
df["email_text"] = df.apply(create_text, axis=1)

print("Example reconstructed emails:")

for text in df["email_text"].head(5):
    print(text[:300])
    print()
```

```python
X_text = df["email_text"]
y = df["Prediction"]

print("Number of emails:", len(X_text))
print("Number of labels:", len(y))
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))
```

```python
vectorizer = CountVectorizer()

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print("Training vector shape:", X_train.shape)
print("Testing vector shape:", X_test.shape)
print("Number of generated features:", X_train.shape[1])
```

```python
model = LogisticRegression(max_iter=1000)

# Measure training time
start_train = time.time()

model.fit(X_train, y_train)

train_time = time.time() - start_train

# Measure prediction time
start_prediction = time.time()

y_pred = model.predict(X_test)

prediction_time = time.time() - start_prediction
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("CountVectorizer + Logistic Regression")
print("--------------------------------------")
print("Accuracy:", accuracy)
print("Number of features:", X_train.shape[1])
print("Training time:", train_time, "seconds")
print("Prediction time:", prediction_time, "seconds")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
improved_vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

X_train_improved = improved_vectorizer.fit_transform(X_train_text)
X_test_improved = improved_vectorizer.transform(X_test_text)

print("Improved training shape:", X_train_improved.shape)
print("Improved testing shape:", X_test_improved.shape)
print("Improved number of features:", X_train_improved.shape[1])
```

```python
improved_model = LogisticRegression(max_iter=1000)

start_train = time.time()

improved_model.fit(X_train_improved, y_train)

improved_train_time = time.time() - start_train

start_prediction = time.time()

y_pred_improved = improved_model.predict(X_test_improved)

improved_prediction_time = time.time() - start_prediction

improved_accuracy = accuracy_score(y_test, y_pred_improved)
```

```python
comparison = pd.DataFrame({
    "Representation": [
        "CountVectorizer",
        "CountVectorizer + Bigrams"
    ],
    "Accuracy": [
        accuracy,
        improved_accuracy
    ],
    "Number of Features": [
        X_train.shape[1],
        X_train_improved.shape[1]
    ],
    "Training Time (seconds)": [
        train_time,
        improved_train_time
    ],
    "Prediction Time (seconds)": [
        prediction_time,
        improved_prediction_time
    ]
})

print(comparison)
```

```python
comparison.to_csv("part_b_c_comparison.csv", index=False)

print("Comparison saved successfully.")
```
