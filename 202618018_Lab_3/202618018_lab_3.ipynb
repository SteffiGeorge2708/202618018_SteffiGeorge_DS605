## Part A - Data Loading and Preprocessing with Scikit-learn
#### Task 1 - Load and Understand the Dataset
● Load hotel_bookings.csv. Use head(), shape, info(), describe(), and dtypes.
● Check the class distribution of is_canceled and use it as the target y.
● Create X from the remaining usable features. Identify numerical and categorical columns.

```python
import pandas as pd

# 1. Load and understand the dataset
df = pd.read_csv("hotel_bookings.csv")

print("Table:")
display(df)

print("Shape:")
print(df.shape)

print("Information:")
df.info()

print("Descriptive statistics:")
display(df.describe())

print("Data types:")
print(df.dtypes)
```

```python
# 2. Check the class distribution of is_canceled and use it as the target y
print("Class distribution of is_canceled:")
print(df["is_canceled"].value_counts())

print("\nClass distribution in percentage:")
print(df["is_canceled"].value_counts(normalize=True) * 100)

y = df["is_canceled"]
```

```python
# 3. Create X from the remaining usable features. Identify numerical and categorical columns
X = df.drop(columns=["is_canceled"])

numerical_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X.select_dtypes(
    include=["object", "string"]
).columns.tolist()

print("Numerical columns:")
print(numerical_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

#### Task 2 - Missing Values, Leakage, and Outliers
● Find the missing-value count and percentage for every column.
● Identify columns with very high missingness. Drop them only if justified and state the reason.
● Remove columns that directly reveal the final booking outcome, such as reservation_status and
reservation_status_date.
● Check selected numerical features for outliers using boxplots and/or the IQR method.
● Remove only clear/extreme outliers and report how many rows were removed.

```python
# 1. Find the missing-value count and percentage for every column
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_summary)
```

```python
# 2. Identify columns with very high missingness
high_missing = missing_summary[
    missing_summary["Missing Percentage"] > 50
]

print("Columns with more than 50% missing values:")
display(high_missing)

# Drop only columns with very high missingness if justified
# In this dataset, 'company' has very high missingness,
# so it can be dropped because most of its values are missing.

X = X.drop(columns=["company"])

print("Reason for dropping 'company':")
print("The company column contains a very high percentage of missing values,")
print("so keeping it would provide very little useful information.")
```

```python
# 3. Remove columns that directly reveal the final booking outcome
# These columns contain information about the final reservation outcome.
X = X.drop(
    columns=["reservation_status", "reservation_status_date"],
    errors="ignore"
)

print("Columns removed due to data leakage:")
print(["reservation_status", "reservation_status_date"])
```

```python
# 4. Check selected numerical features for outliers using the IQR method
selected_numerical = [
    "lead_time",
    "adr",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies"
]

for column in selected_numerical:
    Q1 = X[column].quantile(0.25)
    Q3 = X[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = X[
        (X[column] < lower_bound) |
        (X[column] > upper_bound)
    ]

    print(f"\n{column}:")
    print("Number of outliers:", len(outliers))
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
```

```python
# 5. Remove only clear/extreme outliers and report how many rows were removed
# Use IQR on selected features and remove only extreme values.
rows_before = len(X)

for column in ["lead_time", "adr"]:
    Q1 = X[column].quantile(0.25)
    Q3 = X[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 3 * IQR
    upper_bound = Q3 + 3 * IQR

    X = X[
        (X[column] >= lower_bound) &
        (X[column] <= upper_bound)
    ]

rows_after = len(X)
rows_removed = rows_before - rows_after

# Keep y aligned with X
y = y.loc[X.index]

print("\nRows before outlier removal:", rows_before)
print("Rows after outlier removal:", rows_after)
print("Total rows removed:", rows_removed)
```

#### Task 3 - Create Two Preprocessing Pipelines
● Split the data using train_test_split(test_size=0.2, stratify=y, random_state=42). Use the same split for
all experiments.
● For numerical columns, use KNNImputer(n_neighbors=5) for missing values.
● For categorical columns, use SimpleImputer(strategy="most_frequent") and
OneHotEncoder(handle_unknown="ignore").
● Pipeline A: KNNImputer + StandardScaler for numerical features.
● Pipeline B: KNNImputer + MinMaxScaler for numerical features.
● Use ColumnTransformer and Pipeline so that preprocessing is fitted only on the training data.

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
```

```python
# 1. Split the data using train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


# Identify numerical and categorical columns after Task 2
numerical_cols = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    include=["object", "string"]
).columns.tolist()
```

```python
# 2. Numerical and categorical preprocessing

# Numerical preprocessing
numerical_preprocessing_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

numerical_preprocessing_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# Categorical preprocessing
categorical_preprocessing = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
# 3. Pipeline A - KNNImputer + StandardScaler

preprocessor_A = ColumnTransformer([
    ("numerical", numerical_preprocessing_A, numerical_cols),
    ("categorical", categorical_preprocessing, categorical_cols)
])
```

```python
# 4. Pipeline B - KNNImputer + MinMaxScaler

preprocessor_B = ColumnTransformer([
    ("numerical", numerical_preprocessing_B, numerical_cols),
    ("categorical", categorical_preprocessing, categorical_cols)
])


print("Pipeline A: KNNImputer + StandardScaler")
print(preprocessor_A)

print("\nPipeline B: KNNImputer + MinMaxScaler")
print(preprocessor_B)
```

## Part B - Model Training and Performance Evaluation
#### Task 4 - Train Two Classification Models
● Train LogisticRegression(max_iter=1000) using Pipeline A and Pipeline B.
● Train DecisionTreeClassifier(random_state=42) using Pipeline A and Pipeline B.
● Keep the model settings and train-test split unchanged across the comparison.
● You should obtain four trained model-pipeline combinations in total.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
```

```python
# 1. Logistic Regression using Pipeline A
logistic_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])

# 2. Logistic Regression using Pipeline B
logistic_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])


# 3. Decision Tree using Pipeline A
tree_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", DecisionTreeClassifier(random_state=42))
])


# 4. Decision Tree using Pipeline B
tree_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", DecisionTreeClassifier(random_state=42))
])
```

```python
# Train all four model-pipeline combinations
logistic_A.fit(X_train, y_train)
logistic_B.fit(X_train, y_train)

tree_A.fit(X_train, y_train)
tree_B.fit(X_train, y_train)


print("All four model-pipeline combinations have been trained successfully.")
```

## Task 5 - Evaluate and Compare the Four Results
● For each experiment, calculate training accuracy, testing accuracy, precision, recall, and F1-score.
● Create a final comparison table containing all four model-pipeline combinations.
● Plot a confusion matrix for the best Logistic Regression result and the best Decision Tree result.
● Use the train-test performance difference to comment briefly on possible overfitting.

```python
# Task 5 - Evaluate and Compare the Four Results

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import pandas as pd
import matplotlib.pyplot as plt


# 1. Calculate training and testing performance for each experiment

models = {
    "Logistic Regression + Pipeline A": logistic_A,
    "Logistic Regression + Pipeline B": logistic_B,
    "Decision Tree + Pipeline A": tree_A,
    "Decision Tree + Pipeline B": tree_B
}

results = []

for name, model in models.items():

    # Predictions on training and testing data
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Training accuracy
    train_accuracy = accuracy_score(y_train, y_train_pred)

    # Testing metrics
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred)
    recall = recall_score(y_test, y_test_pred)
    f1 = f1_score(y_test, y_test_pred)

    # Train-test performance difference
    accuracy_difference = train_accuracy - test_accuracy

    results.append({
        "Model-Pipeline": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "Train-Test Difference": accuracy_difference
    })


# 2. Create the final comparison table

comparison_table = pd.DataFrame(results)

display(comparison_table.round(4))


# 3. Find the best Logistic Regression and Decision Tree results

logistic_results = comparison_table[
    comparison_table["Model-Pipeline"].str.contains("Logistic Regression")
]

tree_results = comparison_table[
    comparison_table["Model-Pipeline"].str.contains("Decision Tree")
]

best_logistic_name = logistic_results.loc[
    logistic_results["F1-Score"].idxmax(),
    "Model-Pipeline"
]

best_tree_name = tree_results.loc[
    tree_results["F1-Score"].idxmax(),
    "Model-Pipeline"
]

best_logistic_model = models[best_logistic_name]
best_tree_model = models[best_tree_name]


print("Best Logistic Regression:", best_logistic_name)
print("Best Decision Tree:", best_tree_name)


# 4. Plot confusion matrix for the best Logistic Regression result

y_pred_logistic = best_logistic_model.predict(X_test)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_logistic
)

plt.title(f"Confusion Matrix - {best_logistic_name}")
plt.show()


# Plot confusion matrix for the best Decision Tree result

y_pred_tree = best_tree_model.predict(X_test)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_tree
)

plt.title(f"Confusion Matrix - {best_tree_name}")
plt.show()


# 5. Comment on possible overfitting

print("\nOverfitting Analysis:")

for _, row in comparison_table.iterrows():

    difference = row["Train-Test Difference"]

    if difference > 0.10:
        comment = "Possible overfitting: large difference between training and testing accuracy."
    elif difference > 0.05:
        comment = "Some indication of overfitting: moderate train-test difference."
    else:
        comment = "No major indication of overfitting: train-test difference is relatively small."

    print(f"{row['Model-Pipeline']}: {comment}")
```
