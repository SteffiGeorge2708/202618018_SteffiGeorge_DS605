### Task 2 - Vectorized Arithmetic and Linear Algebra
● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @
or np.matmul().
● For a square matrix, find its transpose and determinant.
● If invertible, find the inverse and verify A @ A^-1 using np.allclose().
● Use vectorized NumPy operations; avoid explicit Python loops.


```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B

print("A + B:")
print(addition)
```

```python
elementwise_multiplication = A * B

print("A * B:")
print(elementwise_multiplication)
```

```python
matrix_multiplication = A @ B

print("A @ B:")
print(matrix_multiplication)

```

```python
# Transpose of A
A_transpose = A.T

print("Transpose of A:")
print(A_transpose)
```

```python
# Calculate determinant of A
det_A = np.linalg.det(A)

print("Determinant of A:", det_A)

# Find inverse of A
A_inverse = np.linalg.inv(A)

print("Inverse of A:")
print(A_inverse)

# Verify A @ A^-1 gives the identity matrix
verification = A @ A_inverse

print("A @ A^-1:")
print(verification)

print("\nIs A @ A^-1 approximately equal to identity matrix?")
print(np.allclose(verification, np.eye(2)))
```

### Task 3 - Normal Distribution and Histogram
● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard
deviation.
● Find the sample mean and sample standard deviation. Compare them with the chosen values.
● Plot a histogram with it.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Chosen mean and standard deviation
mean = 50
std_dev = 10

# Generate 1000 values from a normal distribution
data = np.random.normal(mean, std_dev, 1000)

print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std_dev)
print("Number of Values:", len(data))
```

```python
# Calculate sample mean
sample_mean = np.mean(data)

# Calculate sample standard deviation
sample_std = np.std(data, ddof=1)

print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
```

```python
print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)
print("Difference in Mean:", sample_mean - mean)

print("\nChosen Standard Deviation:", std_dev)
print("Sample Standard Deviation:", sample_std)
print("Difference in Standard Deviation:", sample_std - std_dev)
```

```python
import matplotlib.pyplot as plt

# Create histogram
plt.hist(data, bins=30)

plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

### Task 5 - Filtering and Querying
Use Boolean indexing and/or query() to answer:
● Male passengers older than 50: how many?
● Female first-class passengers: how many, and what percentage survived?
● Age 20-40, Fare above overall median, and survived: how many?
● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
# 1. Male passengers older than 50
male_over_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

print("1. Male passengers older than 50:", len(male_over_50))
```

```python
# Female first-class passengers and percentage survived
female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

total_female_first_class = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", total_female_first_class)
print("   Percentage survived:", survival_percentage)
```

```python
# Age 20-40, Fare above overall median, and survived
overall_median_fare = df["Fare"].median()

condition_3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, above median Fare, and survived:", len(condition_3))
```

```python
# Travelling alone, age < 30, and did not survive
alone_under_30_did_not_survive = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print(
    "\n4. Travelling alone, age < 30, and did not survive:",
    len(alone_under_30_did_not_survive)
)
```

```python
# Embarked S, Pclass 2 or 3, and Fare above Southampton median
southampton_median_fare = df.loc[
    df["Embarked"] == "S",
    "Fare"
].median()

condition_5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print(
    "\n5. Embarked S, Pclass 2 or 3, above Southampton median Fare:",
    len(condition_5)
)
```

### Task 6 - groupby() and Aggregation
● Survival rate by Sex.
● Survival rate by Pclass.
● Average Age and Fare by Pclass.
● Passenger count and survival rate by Sex-Pclass.
● Passenger count, average Fare, and survival rate by Embarked.


```python
# Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("1. Survival rate by Sex:")
print(survival_by_sex)
```

```python
# Survival rate by Pclass
survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival rate by Pclass:")
print(survival_by_pclass)
```

```python
# Average Age and Fare by Pclass
age_fare_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(age_fare_by_pclass)
```

```python
# Passenger count and survival rate by Sex-Pclass
sex_pclass_summary = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("\n4. Passenger count and survival rate by Sex-Pclass:")
print(sex_pclass_summary)
```

```python
# Passenger count, average Fare, and survival rate by Embarked
embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("\n5. Passenger count, average Fare, and survival rate by Embarked:")
print(embarked_summary)
```

### Task 7 - Missing Values and Fare Outliers
● Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
● Fill missing Age values with the mean Age. Show missing counts before and after.
● Try to fill NaN values by imputing using: mean, median, mode, random value
● For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Missing-value count and percentage

missing_count = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing-value count:")
print(missing_count)

print("\nMissing-value percentage:")
print(missing_percentage)


# Plot missing-value counts
missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.show()
```

```python
# 2. Fill missing Age values with mean Age

age_missing_before = df["Age"].isnull().sum()

print("\nMissing Age values before filling:", age_missing_before)

mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

age_missing_after = df["Age"].isnull().sum()

print("Mean Age:", mean_age)
print("Missing Age values after filling:", age_missing_after)
```

```python
# 3. Try different imputation methods

# Mean
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

# Median
df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

# Mode
df_mode = df.copy()
df_mode["Age"] = df_mode["Age"].fillna(df_mode["Age"].mode()[0])

# Random value
df_random = df.copy()

np.random.seed(42)

missing_age = df_random["Age"].isnull()

random_values = np.random.choice(
    df_random.loc[~missing_age, "Age"],
    size=missing_age.sum()
)

df_random.loc[missing_age, "Age"] = random_values

print("\nMissing Age after imputation:")
print("Mean:", df_mean["Age"].isnull().sum())
print("Median:", df_median["Age"].isnull().sum())
print("Mode:", df_mode["Age"].isnull().sum())
print("Random:", df_random["Age"].isnull().sum())
```

```python
# Fare outliers using 1.5 × IQR

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("\nFare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare outliers:", len(fare_outliers))
```

### Task 8 - Features and Pivot Table
● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest
groups.


```python
# Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Create IsAlone
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

# Display the new features
print("New Features:")
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())
```

```python
#Create pivot table
pivot = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("Survival Rate Pivot Table:")
print(pivot)
```

```python
#Clean the table. Remove missing combination
pivot_clean = pivot.dropna()

# Find highest survival group
highest_group = pivot_clean.stack().idxmax()
highest_rate = pivot_clean.stack().max()

# Find lowest survival group
lowest_group = pivot_clean.stack().idxmin()
lowest_rate = pivot_clean.stack().min()

print("\nHighest survival group:")
print("Sex:", highest_group[0])
print("Pclass:", highest_group[1])
print("Survival rate:", highest_rate * 100, "%")

print("\nLowest survival group:")
print("Sex:", lowest_group[0])
print("Pclass:", lowest_group[1])
print("Survival rate:", lowest_rate * 100, "%")
```

### Task 9 - Visualizations and Observations
● Plot a correlation heatmap for relevant numerical columns. Note the strongest positive and negative
relationships.
● Plot the survival rate by Sex and report the male/female survival rates.
● Plot Age vs Fare and distinguish Survived=0 and Survived=1.
● Write 5-7 short observations supported by your numerical results or plots.


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Correlation Heatmap
numeric_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numeric_columns].corr()

print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Titanic Numerical Variables")
plt.show()


# Find strongest correlations
corr_pairs = correlation.where(
    np.triu(np.ones(correlation.shape), k=1).astype(bool)
)

corr_pairs = corr_pairs.stack().sort_values()

print("\nStrongest negative relationships:")
print(corr_pairs.head())

print("\nStrongest positive relationships:")
print(corr_pairs.tail())
```

```python
# 2. Survival Rate by Sex

survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("\nSurvival rates by Sex:")
print(survival_by_sex)

print("\nFemale survival rate:",
      round(survival_by_sex["female"], 2), "%")

print("Male survival rate:",
      round(survival_by_sex["male"], 2), "%")

plt.figure(figsize=(6, 5))

survival_by_sex.plot(kind="bar")

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
# 3. Age vs Fare by Survival

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare by Survival")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

### Observations

1. **Female passengers had a much higher chance of surviving than male passengers.** About **74.20% of females survived**, while only about **18.89% of males survived**.

2. **Passenger class affected the chances of survival.** Passengers in **1st class generally had a higher survival rate** than passengers in 2nd and 3rd class.

3. **Pclass and Fare had a negative relationship.** This is because **1st class has the lower class number**, but passengers in 1st class usually paid higher fares.

4. **SibSp, Parch, and FamilySize are related to each other.** This is because **FamilySize is calculated using SibSp and Parch**, along with the passenger themselves.

5. **Age and Fare do not have a strong relationship.** The graph shows that people of different ages paid different fares, so age does not seem to strongly affect the fare.

6. **Passengers who paid higher fares were often more likely to survive.** However, paying a higher fare did not always mean that a passenger survived.

7. **There is some overlap between survivors and non-survivors in the Age vs Fare graph.** This shows that **age and fare alone cannot clearly tell whether a passenger survived or not**.
