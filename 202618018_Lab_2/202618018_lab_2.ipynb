### Task 2 - Vectorized Arithmetic and Linear Algebra
● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @
or np.matmul().
● For a square matrix, find its transpose and determinant.
● If invertible, find the inverse and verify A @ A^-1 using np.allclose().
● Use vectorized NumPy operations; avoid explicit Python loops.


```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B

print("A + B:")
print(addition)
```

```python
elementwise_multiplication = A * B

print("A * B:")
print(elementwise_multiplication)
```

```python
matrix_multiplication = A @ B

print("A @ B:")
print(matrix_multiplication)

```

```python
# Transpose of A
A_transpose = A.T

print("Transpose of A:")
print(A_transpose)
```

```python
# Calculate determinant of A
det_A = np.linalg.det(A)

print("Determinant of A:", det_A)

# Find inverse of A
A_inverse = np.linalg.inv(A)

print("Inverse of A:")
print(A_inverse)

# Verify A @ A^-1 gives the identity matrix
verification = A @ A_inverse

print("A @ A^-1:")
print(verification)

print("\nIs A @ A^-1 approximately equal to identity matrix?")
print(np.allclose(verification, np.eye(2)))
```

### Task 3 - Normal Distribution and Histogram
● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard
deviation.
● Find the sample mean and sample standard deviation. Compare them with the chosen values.
● Plot a histogram with it.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Chosen mean and standard deviation
mean = 50
std_dev = 10

# Generate 1000 values from a normal distribution
data = np.random.normal(mean, std_dev, 1000)

print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std_dev)
print("Number of Values:", len(data))
```

```python
# Calculate sample mean
sample_mean = np.mean(data)

# Calculate sample standard deviation
sample_std = np.std(data, ddof=1)

print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
```

```python
print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)
print("Difference in Mean:", sample_mean - mean)

print("\nChosen Standard Deviation:", std_dev)
print("Sample Standard Deviation:", sample_std)
print("Difference in Standard Deviation:", sample_std - std_dev)
```

```python
import matplotlib.pyplot as plt

# Create histogram
plt.hist(data, bins=30)

plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

### Task 5 - Filtering and Querying
Use Boolean indexing and/or query() to answer:
● Male passengers older than 50: how many?
● Female first-class passengers: how many, and what percentage survived?
● Age 20-40, Fare above overall median, and survived: how many?
● Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
● Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?

```python
# 1. Male passengers older than 50
male_over_50 = df[
    (df["Sex"] == "male") &
    (df["Age"] > 50)
]

print("1. Male passengers older than 50:", len(male_over_50))
```

```python
# Female first-class passengers and percentage survived
female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

total_female_first_class = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100

print("\n2. Female first-class passengers:", total_female_first_class)
print("   Percentage survived:", survival_percentage)
```

```python
# Age 20-40, Fare above overall median, and survived
overall_median_fare = df["Fare"].median()

condition_3 = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

print("\n3. Age 20-40, above median Fare, and survived:", len(condition_3))
```

```python
# Travelling alone, age < 30, and did not survive
alone_under_30_did_not_survive = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print(
    "\n4. Travelling alone, age < 30, and did not survive:",
    len(alone_under_30_did_not_survive)
)
```

```python
# Embarked S, Pclass 2 or 3, and Fare above Southampton median
southampton_median_fare = df.loc[
    df["Embarked"] == "S",
    "Fare"
].median()

condition_5 = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

print(
    "\n5. Embarked S, Pclass 2 or 3, above Southampton median Fare:",
    len(condition_5)
)
```

### Task 6 - groupby() and Aggregation
● Survival rate by Sex.
● Survival rate by Pclass.
● Average Age and Fare by Pclass.
● Passenger count and survival rate by Sex-Pclass.
● Passenger count, average Fare, and survival rate by Embarked.


```python
# Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("1. Survival rate by Sex:")
print(survival_by_sex)
```

```python
# Survival rate by Pclass
survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100

print("\n2. Survival rate by Pclass:")
print(survival_by_pclass)
```

```python
# Average Age and Fare by Pclass
age_fare_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("\n3. Average Age and Fare by Pclass:")
print(age_fare_by_pclass)
```
